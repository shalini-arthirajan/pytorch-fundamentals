```python
x = t.tensor([
    [1,2,3],
    [4,5,6]
])

print(x.shape)
print(x.ndim)
print(x.dtype)
print(x.size())
print(x.numel())
```

```python
a = t.tensor([1,2,3])
b = t.tensor([4,5,6])

print(a+b)
print(a*b)
#for matrix multiplication
print(a@b) # or t.matmul(a,b) or a.mm(b)
print(a-b)
print(a/b)
```

```python
#for reduction - takes multiple values and combines them into fewer values
x = t.tensor([
    [1,2,3],
    [4,5,6]
])

x.sum()
x.sum(dim=0) #sums down the rows and produces a val for each col
x.sum(dim=1) #sums down the cols and produces a val for each row
```

```python
# broadcasting

'''rules
- start comparing dimensions from the right
- dimensions are compatible if they are equal or one of them is 1
- missing dimensions are treated as 1
- if neither condion is satisfied, an error is raised
'''

a = t.ones((3,1,5)) # valid case
b = t.ones((1,4,5))

c = a + b

print(a.shape)
print(b.shape)
print(c.shape)

# invalid case
#a = t.ones((8,2,6))
#b = t.ones((8,2))

#c = a + b doesnt work since 6 and 2 are not equal therefore incompatible

```

```python
# unsqueeze - inserts a dimension of size 1

x = t.ones((3,5))

print(x.unsqueeze(0).shape)
print(x.unsqueeze(1).shape)
print(x.unsqueeze(2).shape)
```

```python
# squeeze - removes a dimension of size 1

x = t.ones((1,3,1))

print(x.squeeze(0).shape)
print(x.squeeze(1).shape)
print(x.squeeze(2).shape)
```
